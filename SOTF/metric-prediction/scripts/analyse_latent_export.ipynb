```python
import torch

pt_path = "/Users/ricostadeli/Documents/HSG/thesis/ms-thesis/metric-prediction/cache/latent_representation_datasets/ds_1_tune_zoo_cifar10_uniform_large_encoded_checkpoints/token/tune_zoo_cifar10_uniform_large_latent_representations_test_tokens.pt"

data = torch.load(pt_path, map_location="cpu")

def print_structure(obj, prefix=""):
    if isinstance(obj, dict):
        for key, value in obj.items():
            full_key = f"{prefix}.{key}" if prefix else str(key)
            print(full_key)
            print_structure(value, full_key)
    elif hasattr(obj, "columns"):  # pandas DataFrame
        for col in obj.columns:
            print(f"{prefix}.{col}" if prefix else col)
    elif isinstance(obj, (list, tuple)) and len(obj) > 0:
        print_structure(obj[0], f"{prefix}[0]" if prefix else "[0]")

print_structure(data)
```

```python
print(f"Type:  {type(data)}")
print(f"Keys:  {list(data.keys())}\n")

for key, val in data.items():
    if isinstance(val, torch.Tensor):
        print(f"  {key!r:25s}  Tensor   shape={tuple(val.shape)}  dtype={val.dtype}")
    elif isinstance(val, np.ndarray):
        print(f"  {key!r:25s}  ndarray  shape={val.shape}  dtype={val.dtype}")
    elif isinstance(val, list):
        first = val[0] if val else None
        inner = type(first).__name__ if first is not None else "?"
        if isinstance(first, torch.Tensor):
            inner = f"Tensor shape={tuple(first.shape)}"
        elif isinstance(first, np.ndarray):
            inner = f"ndarray shape={first.shape}"
        print(f"  {key!r:25s}  list[{inner}]  len={len(val)}")
    else:
        preview = repr(val)
        if len(preview) > 80:
            preview = preview[:80] + "..."
        print(f"  {key!r:25s}  {type(val).__name__}  = {preview}")

# ── per-key detail ────────────────────────────────────────────────────────────
print("\n── mode ──────────────────────────────────────────────────────────────────")
print(f"  value: {data['mode']!r}")

print("\n── model_ids (first 5) ───────────────────────────────────────────────────")
ids = data["model_ids"]
ids_list = ids.tolist() if isinstance(ids, (torch.Tensor, np.ndarray)) else list(ids)
print(f"  total: {len(ids_list)}")
for mid in ids_list[:5]:
    print(f"    {mid!r}")

print("\n── checkpoint_paths (first 5) ────────────────────────────────────────────")
cp = data["checkpoint_paths"]
cp_list = cp.tolist() if isinstance(cp, (torch.Tensor, np.ndarray)) else list(cp)
print(f"  total: {len(cp_list)}")
for p in cp_list[:5]:
    print(f"    {p!r}")

print("\n── epoch ─────────────────────────────────────────────────────────────────")
ep = data["epoch"]
ep_arr = ep.numpy() if isinstance(ep, torch.Tensor) else np.asarray(ep)
print(f"  shape: {ep_arr.shape}  dtype: {ep_arr.dtype}")
print(f"  unique values ({len(np.unique(ep_arr))}): {np.unique(ep_arr)}")
print(f"  first 10: {ep_arr[:10]}")

print("\n── embeddings ────────────────────────────────────────────────────────────")
emb = data["embeddings"]

def _to_f32(t):
    return t.detach().cpu().float().numpy() if isinstance(t, torch.Tensor) else np.asarray(t, dtype=np.float32)

if isinstance(emb, list):
    token_counts = [e.shape[0] for e in emb]
    latent_dim = emb[0].shape[-1]
    print(f"  list of {len(emb)} entries  (one per checkpoint)")
    print(f"  latent_dim:   {latent_dim}")
    print(f"  token counts — min: {min(token_counts)}  max: {max(token_counts)}  "
          f"mean: {np.mean(token_counts):.0f}  unique shapes: {len(set(token_counts))}")
    # mean-pool over token dim → (n_samples, latent_dim) for aggregate stats
    emb_pooled = np.stack([_to_f32(t).mean(axis=0) for t in emb])
    print(f"\n  mean-pooled → {emb_pooled.shape}  (n_samples × latent_dim)")
else:
    emb_pooled = _to_f32(emb)
    print(f"  shape: {emb_pooled.shape}  (n_samples × latent_dim)")

print(f"  dtype:              {emb_pooled.dtype}")
print(f"  NaN / Inf:          {np.isnan(emb_pooled).sum()} / {np.isinf(emb_pooled).sum()}")
print(f"  global mean:        {emb_pooled.mean():.4f}")
print(f"  global std:         {emb_pooled.std():.4f}")
print(f"  per-dim std range:  [{emb_pooled.std(axis=0).min():.4f}, {emb_pooled.std(axis=0).max():.4f}]")
dead_dims = int((emb_pooled.std(axis=0) < 1e-6).sum())
print(f"  near-zero-std dims: {dead_dims} / {emb_pooled.shape[-1]}")
```

```python
emb = data["embeddings"]
shapes = [tuple(t.shape) for t in emb]
unique_shapes = set(shapes)

print(f"Total entries: {len(shapes)}")
print(f"Unique shapes: {unique_shapes}")
if len(unique_shapes) == 1:
    print("All entries have the same shape — embeddings are uniform.")
else:
    from collections import Counter
    counts = Counter(shapes)
    print(f"Shape distribution ({len(unique_shapes)} distinct shapes):")
    for shape, count in sorted(counts.items()):
        print(f"  {shape}: {count} entries")
```
